# Como se entrena una regresión lineal simple


En toda operación de regression lo que buscamos es minimizar el ruido (que esperamos que sea gaussiano), que se refiere a que tan alejado se encuentran los puntos de nuestra estimación lineal.

Para poder determinar cuál es el mejor caso o el mejor ajuste de parámetros tenemos una función de perdida 
$$l(w,b) = \frac{1}{N}\sum_{n=1}^{N} (y - (w\times x + b))²$$

Qué será la funciṕon que se busca minimizar, el mejor ajuste del modelo será aquel que minimize esta pérdida. A esto nos referimos por mínimos cuadrados.

Ahora, para determinar cuál deberían ser estos valores, si analizamos la función de perdida solo en relación al peso, tendremos $l(w) = (y - w\times x)$
Que resulta ser una función cuadrática, y esto es una parábola, que podremos optimizar y encontrar su punto mínimo, a través de su derivada, claro, esto no se podrá hacer en modelos más complejos, pero marca el camino de como se puede encontrar ese weight optimo que minimize la perdida.

## Gradiente Descendiente
El gradiente descendiente es un método iterativo que nos permite poder estimar el peso optimo que minimiza la función de pérdida
Se compone de la siguiente forma
1. Toma un valor aleatorio $w_0$
2. Luego calcula $w_1 = w_0 \times \eta \frac{dl(w_0)}{dw}$
3. Así sucesivamente $w_2 = w_1 \times \eta \frac{dl(w_1)}{dw}$

Donde $\eta$ se denomina el Learning Rate, este vive entre 0 y 1. Y determina la velocidad o el tamaño de los saltos que ira haciendo el gradiente descendiente para encontrar el punto opimo,

### Cuál es el learning rate optimo
No existe un método o valor optimo, si escogemos un valor muy grande, puede hacer que el gradiente salte entre los extremos de la parábola y nunca minimize la perdida. Mientras que un valor muy pequeño puede hacer que la iteración de pasos muy pequeños y le cueste mucho llegar.

### ¿Cómo sabemos si debemos parar entonces el Gradiente?
El método más utilizado para saber cuando detenerlo, es en el momento que el algoritmo deja de optimizar la pérdida.

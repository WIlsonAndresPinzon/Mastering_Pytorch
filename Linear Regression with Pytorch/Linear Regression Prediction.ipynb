# Regresion Lineal
Ya la conocemos, se define como $$\hat{y} = b + w \times x$$

Donde tenemos el error o bias $b$ y el peso $w$

```python
w = torch.tensor(2.0, requires_grad = True)
b = torch.tensor(-1.0, requires_grad = True)
```

```python
def forward(x):
    y = b + w*x
    return y
```

```python
x = torch.tensor(1.0)
```

```python
yhat = forward(x)
print(yhat)
```

```python
x = torch.tensor([[1.0],[2.0]])
yhat = forward(x)
print(yhat)
```

```python
from torch.nn import Linear
```

```python
torch.manual_seed(1)

model = Linear(in_features = 1, out_features = 1)
```

```python
print(list(model.parameters()))
```

```python
yhat = model(x)
yhat
```

## Custom Modules
Básicamente es un objeto que nos permite crear modelos más complejos desde objetos ya de torch

```python
import torch.nn as nn
```

```python
class LR(nn.Module):
    def __init__(self, in_features, out_features):
        super(LR,self).__init__()
        self.linear = nn.Linear(in_features = in_features, out_features = out_features)

    def forward(self, x):
        out = self.linear(x)
        return out
```

```python
model = LR(1,1) 
```

```python
model(x)
```

```python
model.state_dict()
```

```python
model.state_dict()['linear.weight'].data[0] = torch.tensor([0.5153])
model.state_dict()['linear.bias'].data[0] = torch.tensor([-0.44])
```

```python
model(x)
```
